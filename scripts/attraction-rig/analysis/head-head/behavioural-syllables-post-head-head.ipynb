```python

import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
from itertools import combinations
from scipy.stats import kruskal, mannwhitneyu
from statsmodels.stats.multitest import multipletests
from IPython.display import display
import numpy as np


df1 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/fed-fed/behaviour_detection.csv')


df1_contacts = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/fed-fed/head_head_contacts_kinematics_over_time_nocontacts_ref-window10.csv')
df1_contacts = df1_contacts[
    (df1_contacts['interaction_number'] == 1) &
    (df1_contacts['interaction_frame'] < 1200)
].copy()

df1_contacts["file"] = (
    df1_contacts["file"]
    .astype(str)
    .str.replace(".tracks.feather", "", regex=False)
)

df1_contacts = df1_contacts.merge(
    df1[["file", "frame", 'track_id', "behaviour"]],
    on=["file", "frame", "track_id"],
    how="left"
)
df1_contacts['Condition'] = 'GH Fed-Fed'

df2 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/fed-starved/behaviour_detection.csv')


df2_contacts = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/fed-starved/head_head_contacts_kinematics_over_time_nocontacts_ref-window10.csv')

df2_contacts = df2_contacts[
    (df2_contacts['interaction_number'] == 1) &
    (df2_contacts['interaction_frame'] < 1200)
].copy()





df2_contacts["file"] = (
    df2_contacts["file"]
    .astype(str)
    .str.replace(".tracks.feather", "", regex=False)
)

df2_contacts = df2_contacts.merge(
    df2[["file", "frame", 'track_id', "behaviour"]],
    on=["file", "frame", "track_id"],
    how="left"
)

df2_contacts['Condition'] = 'GH Fed-Starved'


df3 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/starved-starved/behaviour_detection.csv')


df3_contacts = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/starved-starved/head_head_contacts_kinematics_over_time_nocontacts_ref-window10.csv')

df3_contacts = df3_contacts[
    (df3_contacts['interaction_number'] == 1) &
    (df3_contacts['interaction_frame'] < 1200)
].copy()

df3_contacts["file"] = (
    df3_contacts["file"]
    .astype(str)
    .str.replace(".tracks.feather", "", regex=False)
)

df3_contacts = df3_contacts.merge(
    df3[["file", "frame", 'track_id', "behaviour"]],
    on=["file", "frame", "track_id"],
    how="left"
)

df3_contacts['Condition'] = 'GH Starved-Starved'

df4 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/fed-fed/behaviour_detection.csv')

df4_contacts = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/fed-fed/head_head_contacts_kinematics_over_time_nocontacts_ref-window10.csv')

df4_contacts = df4_contacts[
    (df4_contacts['interaction_number'] == 1) &
    (df4_contacts['interaction_frame'] < 1200)
].copy()

df4_contacts["file"] = (
    df4_contacts["file"]
    .astype(str)
    .str.replace(".tracks.feather", "", regex=False)
)

df4_contacts = df4_contacts.merge(
    df4[["file", "frame",'track_id',  "behaviour"]],
    on=["file", "frame", "track_id"],
    how="left"
)

df4_contacts['Condition'] = 'SI Fed-Fed'

df5 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/fed-starved/behaviour_detection.csv')


df5_contacts = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/fed-starved/head_head_contacts_kinematics_over_time_nocontacts_ref-window10.csv')

df5_contacts = df5_contacts[
    (df5_contacts['interaction_number'] == 1) &
    (df5_contacts['interaction_frame'] < 1200)
].copy()

df5_contacts["file"] = (
    df5_contacts["file"]
    .astype(str)
    .str.replace(".tracks.feather", "", regex=False)
)

df5_contacts = df5_contacts.merge(
    df5[["file", "frame", 'track_id', "behaviour"]],
    on=["file", "frame", "track_id"],
    how="left"
)

df5_contacts['Condition'] = 'SI Fed-Starved'

df6 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/starved-starved/behaviour_detection.csv')

df6_contacts = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/starved-starved/head_head_contacts_kinematics_over_time_nocontacts_ref-window10.csv')

df6_contacts = df6_contacts[
    (df6_contacts['interaction_number'] == 1) &
    (df6_contacts['interaction_frame'] < 1200)
].copy()

df6_contacts["file"] = (
    df6_contacts["file"]
    .astype(str)
    .str.replace(".tracks.feather", "", regex=False)
)

df6_contacts = df6_contacts.merge(
    df6[["file", "frame", 'track_id', "behaviour"]],
    on=["file", "frame", "track_id"],
    how="left"
)

df6_contacts['Condition'] = 'SI Starved-Starved'

df = pd.concat([df1_contacts, df2_contacts, df3_contacts, df4_contacts, df5_contacts, df6_contacts], ignore_index=True)

df = df.sort_values(['Condition', "file", "track_id", "frame"]).copy()

df["prev_behaviour"] = (
    df.groupby(['Condition', "file", "track_id"])["behaviour"]
    .shift()
)

df["is_initiation"] = (
    df["prev_behaviour"].notna() &
    df["behaviour"].notna() &
    (df["behaviour"] != df["prev_behaviour"])
)



df["contact_period"] = np.where(
    df["rel_frame"] < 0,
    "Before contact",
    "After contact"
)


si = df[df["Condition"].str.startswith("SI")].copy()
gh = df[df["Condition"].str.startswith("GH")].copy()

# Optional: remove SI/GH prefix so plot labels are cleaner
si["feeding_condition"] = si["Condition"].str.replace("SI ", "", regex=False)
gh["feeding_condition"] = gh["Condition"].str.replace("GH ", "", regex=False)
```

```python
gh_bout_counts = (
    gh[gh["is_initiation"]]
    .groupby(["file", "Condition", "contact_period", "behaviour"])
    .size()
    .reset_index(name="n_bouts")
)

condition_order = ["GH Fed-Fed", "GH Fed-Starved", "GH Starved-Starved"]
period_order = ["Before contact", "After contact"]

for condition in condition_order:
    plot_df = gh_bout_counts[gh_bout_counts["Condition"] == condition].copy()

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=plot_df,
        x="behaviour",
        y="n_bouts",
        hue="contact_period",
        hue_order=period_order,
        errorbar="sd"
    )

    plt.title(f"{condition}: behaviour initiations before vs after contact")
    plt.xlabel("Behaviour")
    plt.ylabel("Number of bouts / initiations per file")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```
