```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import sys
import matplotlib.patches as mpatches
import matplotlib as mpl

# ---- Adobe-friendly fonts (must be set BEFORE plotting) ----
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42



df1 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/fed-fed/digging_behaviour.csv')
df1['Condition'] = 'GH Fed-Fed Agarose'

df2 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/fed-starved/digging_behaviour.csv')
df2['Condition'] = 'GH Fed-Starved Agarose'

df3 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/group-housed/starved-starved/digging_behaviour.csv')
df3['Condition'] = 'GH Starved-Starved Agarose'

df4 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/fed-fed/digging_behaviour.csv')
df4['Condition'] = 'SI Fed-Fed Agarose'

df5 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/fed-starved/digging_behaviour.csv')
df5['Condition'] = 'SI Fed-Starved Agarose'

df6 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/agarose-plates/socially-isolated/starved-starved/digging_behaviour.csv')
df6['Condition'] = 'SI Starved-Starved Agarose'

df7 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/food-plates/group-housed/fed-fed/digging_behaviour.csv')
df7['Condition'] = 'GH Fed-Fed Food'

df8 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/food-plates/group-housed/fed-starved/digging_behaviour.csv')
df8['Condition'] = 'GH Fed-Starved Food'

df9 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/food-plates/group-housed/starved-starved/digging_behaviour.csv')
df9['Condition'] = 'GH Starved-Starved Food'

df10 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/food-plates/socially-isolated/fed-fed/digging_behaviour.csv')
df10['Condition'] = 'SI Fed-Fed Food'

df11 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/food-plates/socially-isolated/fed-starved/digging_behaviour.csv')
df11['Condition'] = 'SI Fed-Starved Food'

df12 = pd.read_csv('/Volumes/lab-windingm/home/users/cochral/LRS/AttractionRig/analysis/social-isolation/head-head/2/food-plates/socially-isolated/starved-starved/digging_behaviour.csv')
df12['Condition'] = 'SI Starved-Starved Food'



df = pd.concat([df1, df2, df3, df4, df5, df6, df7, df8, df9, df10, df11, df12], ignore_index=True)



```

```python
plot_df = (
    df
    .groupby("Condition")
    .agg(
        n_yes=("digging_behaviour", lambda x: (x == "Y").sum()),
        total=("digging_behaviour", "count")
    )
    .reset_index()
)

plot_df["percentage_yes"] = (plot_df["n_yes"] / plot_df["total"]) * 100

condition_order = [
    "GH Fed-Fed Agarose",
    "GH Fed-Starved Agarose",
    "GH Starved-Starved Agarose",
    "SI Fed-Fed Agarose",
    "SI Fed-Starved Agarose",
    "SI Starved-Starved Agarose",

    "GH Fed-Fed Food",
    "GH Fed-Starved Food",
    "GH Starved-Starved Food",
    "SI Fed-Fed Food",
    "SI Fed-Starved Food",
    "SI Starved-Starved Food",
]

plt.figure(figsize=(7, 5))

sns.barplot(
    data=plot_df,
    x="Condition",
    y="percentage_yes",
    order=condition_order

)

plt.ylabel("Percentage of videos (%)")
plt.xlabel("Condition")
plt.title("Digging behaviour by condition")
plt.ylim(0, 100)

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
